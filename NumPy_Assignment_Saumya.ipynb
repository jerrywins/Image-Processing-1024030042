```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
array1 = np.array([[1, 2, 3],
                   [2, 4, 5],
                   [1, 2, 3]])

print("Using flatten():")
print(array1.flatten())

print("Using ravel():")
print(array1.ravel())
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[1, 2], [3, 4]])

print("Element-wise comparison:")
print(arr1 == arr2)

print("Are the arrays equal?")
print(np.array_equal(arr1, arr2))
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)

i = np.argmax(counts)
value = values[i]

print("Most frequent value:", value)
print("Frequency:", counts[i])
print("Indices:", np.where(x == value)[0])
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3])

values, counts = np.unique(y, return_counts=True)

i = np.argmax(counts)
value = values[i]

print("Most frequent value:", value)
print("Frequency:", counts[i])
print("Indices:", np.where(y == value)[0])
```

```python
gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]')

print("Matrix:")
print(gfg)

print("Sum of all elements:", np.sum(gfg))
print("Row-wise sum:", np.sum(gfg, axis=1))
print("Column-wise sum:", np.sum(gfg, axis=0))
```

```python
n_array = np.array([[55, 25, 15],
                    [30, 44, 2],
                    [11, 45, 77]])

print("Matrix:")
print(n_array)

print("\nSum of diagonal elements:")
print(np.trace(n_array))

eigenvalues, eigenvectors = np.linalg.eig(n_array)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

print("\nInverse:")
print(np.linalg.inv(n_array))

print("\nDeterminant:")
print(np.linalg.det(n_array))
```

```python
p = np.array([[1, 2],
              [2, 3]])

q = np.array([[4, 5],
              [6, 7]])

print("Matrix multiplication:")
print(np.matmul(p, q))

print("\nCovariance:")
print(np.cov(p.flatten(), q.flatten()))
```

```python
p = np.array([[1, 2],
              [2, 3],
              [4, 5]])

q = np.array([[4, 5, 1],
              [6, 7, 2]])

print("Matrix multiplication:")
print(np.matmul(p, q))

print("\nCovariance:")
print(np.cov(p.flatten(), q.flatten()))
```

```python
x = np.array([[2, 3, 4],
              [3, 2, 9]])

y = np.array([[1, 5, 0],
              [5, 10, 3]])

print("Inner product:")
print(np.inner(x, y))

print("\nOuter product:")
print(np.outer(x, y))

print("\nCartesian product:")
print(np.array(np.meshgrid(x.flatten(), y.flatten())).T.reshape(-1, 2))
```

```python
array = np.array([[1, -2, 3],
                  [-4, 5, -6]])

print("Array:")
print(array)
```

```python
print("Absolute value:")
print(np.abs(array))
```

```python
print("Flattened array percentiles:")
print(np.percentile(array, [25, 50, 75]))
```

```python
print("Column-wise percentiles:")
print(np.percentile(array, [25, 50, 75], axis=0))
```

```python
print("Row-wise percentiles:")
print(np.percentile(array, [25, 50, 75], axis=1))
```

```python
print("Flattened array:")

print("Mean:", np.mean(array))
print("Median:", np.median(array))
print("Standard deviation:", np.std(array))
```

```python
print("Column-wise:")

print("Mean:", np.mean(array, axis=0))
print("Median:", np.median(array, axis=0))
print("Standard deviation:", np.std(array, axis=0))
```

```python
print("Row-wise:")

print("Mean:", np.mean(array, axis=1))
print("Median:", np.median(array, axis=1))
print("Standard deviation:", np.std(array, axis=1))
```

```python
from PIL import Image
import numpy as np
```

```python
def img_to_array(path):
    img = Image.open(path)
    arr = np.array(img)

    if len(arr.shape) == 2:
        np.savetxt("image_gray.txt", arr, fmt="%d")
        print("Grayscale image saved as image_gray.txt")
    else:
        np.savetxt("image_rgb.txt", arr.reshape(arr.shape[0], -1), fmt="%d")
        print("RGB image saved as image_rgb.txt")

    return arr
```

```python
arr = img_to_array("20221230_160548.jpg")
```

```python
print("Shape:", arr.shape)
```
